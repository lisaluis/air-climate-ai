```
import os, subprocess
REPO = "https://github.com/lisaluis/air-climate-ai"
if not os.path.exists("repo"): subprocess.run(["git", "clone", "-q", REPO, "repo"], check=True)
%cd repo
from google.colab import userdata
os.environ["GROQ_API_KEY"] = userdata.get("GROQ_API_KEY")      # never print or commit the key
```

```
import threading
from werkzeug.serving import make_server
import app as web
server = make_server("127.0.0.1", 5000, web.app); threading.Thread(target=server.serve_forever, daemon=True).start()
print("running")
```
